```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('D:/Amazon Customer Analysis/amazon_ecommerce_1M.csv')
```

```python
df.head()
```

```python
df['purchase_date'] = pd.to_datetime(df['purchase_date'], format = '%d-%m-%Y')
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['rating'] = df.groupby(['category','subcategory', 'brand'])['rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df['seller_rating'] = df.groupby(['category','subcategory','brand'])['seller_rating'].transform(lambda x: x.fillna(x.median()))
```
